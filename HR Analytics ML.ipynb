```python
from sklearn.metrics import make_scorer, f1_score, roc_auc_score
```

```python
# Define scorer
def f1_metric(y_test, y_pred):
    score = f1_score(y_test, y_pred)
    return score
```

```python
f1_scorer = make_scorer(f1_metric,greater_is_better=True)
```

```python
# Applying Grid Search to find the best model and the best parameters
from sklearn.model_selection import GridSearchCV
parameters = {"n_estimators" : [150,200,300,450,500],
              "criterion":['gini','entropy'],
              "max_depth": [3,5,7,10],
              "min_samples_split": [50, 30],
              "min_samples_leaf": [1, 5, 15],
              "min_weight_fraction_leaf": [0.1, 0.0025, 0.005],
              #"splitter": ['best','random']
              }

grid_search = GridSearchCV(estimator = rfc,
                           param_grid = parameters,
                           scoring = f1_scorer,
                           cv = 10,
                           n_jobs = -1)
grid_search = grid_search.fit(X_train, y_train)
```

```python
best_parameters = grid_search.best_params_
best_parameters
```

```python
best_accuracy = grid_search.best_score_
best_accuracy
```

```python
# Fitting Final Model on training set
from sklearn.tree import DecisionTreeClassifier

tunedDTC=DecisionTreeClassifier (n_estimators = best_parameters["n_estimators"]
                                 criterion = best_parameters["criterion"],
                                 splitter = best_parameters["splitter"],
                                 max_depth = best_parameters["max_depth"],
                                 min_samples_split = best_parameters["min_samples_split"],
                                 min_samples_leaf = best_parameters["min_samples_leaf"],
                                 min_weight_fraction_leaf=best_parameters["min_weight_fraction_leaf"]
                                 )
tunedDTC.fit(X_train, y_train)
```

```python
print('Confusion Matrix for Tuned RF: \n',confusion_matrix(y_test, tunedDTC.predict(X_test)))
print('Accuracy for Tuned RF: \n',accuracy_score(y_test, tunedDTC.predict(X_test)))
acc.append(accuracy_score(y_test, tunedDTC.predict(X_test)))
print('Precision for Tuned RF: \n',precision_score(y_test, tunedDTC.predict(X_test)))
precision.append(precision_score(y_test, tunedDTC.predict(X_test)))
print('Recall for Tuned RF: \n',recall_score(y_test, tunedDTC.predict(X_test)))
recall.append(recall_score(y_test, tunedDTC.predict(X_test)))
print('f1_score for Tuned RF: \n',f1_score(y_test, tunedDTC.predict(X_test)))
f1.append(f1_score(y_test, tunedDTC.predict(X_test)))
```
